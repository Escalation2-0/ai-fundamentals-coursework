# **Operators**

1. Arthmetic Operators

```python
# Addition
print("10 + 3 = " + str(10 + 3))

# Subtraction
print("10 - 3 = " + str(10 - 3))

# Multiplication
print("10 * 3 = " + str(10 * 3))

# Division
print("10 / 3 = " + str(10 / 3))

# Division with integer
print("10 //3 = " + str(10 // 3))

# Modulus
print("10 % 3 = " + str(10 % 3))

# Exponent
print("10 ** 3 = " + str(10 ** 3))
```

```python
# Augmented operators
x = 10
x += 3
print(x)
```

Operator Precedence

```python
x = 10 + 3 * 2
print (x)
```

2. Comparison Operators

```python
# They return boolean (True/False) solutions

# Greater than
print("10 > 3: " + str(10 > 3))
# Greater than or equal to
print("3 >= 3: " + str(10 >= 3))
# Less than
print("10 < 3: " + str(10 < 3))
# Less than or equal to
print("10 <= 3: " + str(10 <= 3))
# Equal
print("10 == 3: " + str(10 == 3))
# Not equal
print("10 != 3: " + str(10 != 3))
```

4. Logical Operators

```python
# Used for building complex rules and conditions

# Using "and" check that a price is between two values
price = 25
print("Price between 10 and 30: " + str(price > 10 and price < 30))

# Using "or" check that a price is either of two values
price = 25
print("Price is either 10 or 30: " + str(price == 10 or price == 30))

# Using "not" check that a price is less than a value
price = 25
print("Price is less than 30: " + str(not price > 30))
```

**Exercise**

* Get weight as input
* Get unit as input (Kgs or Lbs)
* Display weight converted to the opposite unit



```python
# ask for weight and unit, then convert to the other unit
weight = float(input("Enter your weight: "))
unit = input("Enter unit (Kgs or Lbs): ")

# .upper() makes the answer uppercase so "lbs" and "Lbs" both work
if unit.upper() == "KGS" or unit.upper() == "KG":
    # kgs to lbs
    converted = weight * 2.2
    print("Weight in Lbs: " + str(converted))
elif unit.upper() == "LBS" or unit.upper() == "LB":
    # lbs to kgs
    converted = weight / 2.2
    print("Weight in Kgs: " + str(converted))
else:
    print("Unknown unit. Please use Kgs or Lbs.")
```

List methods

```python
numbers = [1, 2, 3, 4, 5, 6]
# Append — add a number at the end
numbers.append(7)
print(numbers)

# Insert — add a number at the start
numbers.insert(0, 0)
print(numbers)

# Remove — take out the number 3
numbers.remove(3)
print(numbers)

# Clear — wipe the whole list
numbers.clear()
print(numbers)

# make the list again for the last demos
numbers = [1, 2, 3, 4, 5, 6]
# In — is 10 in the list?
print(10 in numbers)
# Len — how many items are in the list?
print (len(numbers))
```
